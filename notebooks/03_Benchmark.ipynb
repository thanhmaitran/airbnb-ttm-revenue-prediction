# 03 — Benchmark: Comparing All Four Models

Runs the same routine as `benchmark.py` (via `src/training_utils.py`) and visualises the comparison. For the actual CLI (writes `reports/tables/benchmark_results.csv` and saves fitted pipelines), use:

```bash
python benchmark.py
```

````python
import sys
sys.path.insert(0, "..")

import matplotlib.pyplot as plt
import pandas as pd

from src.training_utils import load_training_frame, train_one_model
from src.utils.seed import set_global_seed
from src.utils.config import load_training_config

set_global_seed(load_training_config()["seed"])
X_train, X_test, y_train, y_test = load_training_frame()
X_train.shape, X_test.shape
````

## Train and score all four models

````python
results = []
for model_name in ["ridge", "hist_gb", "random_forest", "mlp"]:
    print(f"Training {model_name}...")
    r = train_one_model(model_name, X_train, X_test, y_train, y_test)
    results.append(r)

table = pd.DataFrame(results)[["model", "MAE", "RMSE", "R2_test", "R2_cv", "R2_cv_std", "time_seconds"]]
table.sort_values("R2_test", ascending=False)
````

## Visualise MAE / RMSE / R² across models

Reproduces the shape of Figure 9 in `reports/technical_report.md`.

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for ax, metric, title in zip(axes, ["MAE", "RMSE", "R2_test"],
                              ["MAE (USD) — lower is better", "RMSE (USD) — lower is better", "R² — higher is better"]):
    ax.bar(table["model"], table[metric], color=["#4C72B0", "#DD8452", "#55A868", "#8172B2"])
    ax.set_title(title)
    ax.tick_params(axis="x", rotation=30)
plt.tight_layout()
plt.show()
````
